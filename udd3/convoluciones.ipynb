```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

random_img=np.random.randint(0,256, size=(256,256), dtype=np.uint8)
img_gauss=cv2.GaussianBlur(random_img,(5,5),0)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title('Imagen Original')
plt.imshow(random_img,cmap='gray')
plt.subplot(1,2,2)
plt.title('Imagen con Filtro Gaussiano')
plt.imshow(img_gauss,cmap='gray')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def convolucion_media_h(imagen, tkernel):
    kernel = np.zeros((tkernel, tkernel), np.float32)
    kernel[tkernel//2, :] = 1
    kernel = kernel / tkernel
    return cv2.filter2D(imagen, -1, kernel)
 
def convolucion_media_v(imagen, tkernel):
    kernel = np.zeros((tkernel, tkernel), np.float32)
    kernel[:, tkernel//2] = 1
    kernel = kernel / tkernel
    return cv2.filter2D(imagen, -1, kernel)
def convolucion_media_diag(imagen, tkernel):
    kernel = np.zeros((tkernel, tkernel), np.float32)
    np.fill_diagonal(kernel, 1)
    kernel = kernel / tkernel
    return cv2.filter2D(imagen, -1, kernel)
def convolucion_media_diag_inv(imagen, tkernel):
    kernel = np.fliplr(np.eye(tkernel, dtype=np.float32))
    kernel = kernel / tkernel
    return cv2.filter2D(imagen, -1, kernel)
img = cv2.imread("paisaje.jpg", 0)
k = 80
 
img_h = convolucion_media_h(img, k)
img_v = convolucion_media_v(img, k)
img_d = convolucion_media_diag(img, k)
img_di = convolucion_media_diag_inv(img, k)
plt.figure(figsize=(12,10))
 
plt.subplot(231)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis('off')
 
plt.subplot(232)
plt.imshow(img_h, cmap='gray')
plt.title("Horizontal FUERTE")
plt.axis('off')
 
plt.subplot(233)
plt.imshow(img_v, cmap='gray')
plt.title("Vertical FUERTE")
plt.axis('off')
 
plt.subplot(234)
plt.imshow(img_d, cmap='gray')
plt.title("Diagonal ↘ FUERTE")
plt.axis('off')
 
plt.subplot(235)
plt.imshow(img_di, cmap='gray')
plt.title("Diagonal ↙ FUERTE")
plt.axis('off')
 
plt.tight_layout()
plt.show()
```
