moire

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('moire.jpg', cv2.IMREAD_GRAYSCALE)

limpia_suave = cv2.GaussianBlur(img, (5, 5), 0)

limpia_mediana = cv2.medianBlur(img, 5)

limpia_bilateral = cv2.bilateralFilter(img, 9, 75, 75)

plt.figure(figsize=(15, 5))

plt.subplot(141)
plt.imshow(img, cmap='gray')
plt.title('Con Moiré')
plt.axis('off')

plt.subplot(142)
plt.imshow(limpia_suave, cmap='gray')
plt.title('Filtro Gaussiano')
plt.axis('off')

plt.subplot(143)
plt.imshow(limpia_mediana, cmap='gray')
plt.title('Filtro Mediana')
plt.axis('off')

plt.subplot(144)
plt.imshow(limpia_bilateral, cmap='gray')
plt.title('Filtro Bilateral')
plt.axis('off')

plt.tight_layout()
plt.show()
```

aliasing

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('aliasing.jpg', cv2.IMREAD_GRAYSCALE)

limpia_gauss = cv2.GaussianBlur(img, (9, 9), 0)

limpia_bilateral = cv2.bilateralFilter(img, 15, 75, 75)

limpia_fuerte = cv2.GaussianBlur(img, (17, 17), 0)
plt.figure(figsize=(15, 5))

plt.subplot(141)
plt.imshow(img, cmap='gray')
plt.title('Imagen Original (Con Aliasing)')
plt.axis('off')

plt.subplot(142)
plt.imshow(limpia_gauss, cmap='gray')
plt.title('Filtro Gaussiano (Anti-aliasing)')
plt.axis('off')

plt.subplot(143)
plt.imshow(limpia_bilateral, cmap='gray')
plt.title('Filtro Bilateral')
plt.axis('off')

plt.subplot(144)
plt.imshow(limpia_fuerte, cmap='gray')
plt.title('Antialiasing Fuerte')
plt.axis('off')

plt.tight_layout()
plt.show()
```

Aliasing:
El aliasing es un problema que aparece cuando una señal continua (como una imagen, un sonido o un modelo 3D) se convierte a formato digital pero no se toman suficientes muestras para representarla bien. En imágenes se nota clarísimo: los bordes que deberían ser suaves se ven como en forma de escalera, con picos o dientes, sobre todo en líneas diagonales o curvas. Esto pasa porque el sistema no tiene la resolución suficiente para capturar todos los detalles, entonces “inventa” una versión más simple que genera esos errores visuales. También puede pasar en audio, donde se escuchan distorsiones raras. Es muy común en gráficos por computadora, videojuegos y renderizado. Para solucionarlo se usa el antialiasing, que lo que hace es suavizar los bordes mezclando colores entre píxeles para que todo se vea más continuo y realista.
Conclusión: El aliasing es básicamente una pérdida de calidad por falta de detalle al digitalizar, y el antialiasing es clave para mejorar la apariencia y hacer que las imágenes se vean mucho más suaves y profesionales.

Moiré:
El efecto moiré es un fenómeno visual que ocurre cuando se superponen dos patrones repetitivos, como líneas o puntos, y se crea un patrón nuevo que no estaba en la imagen original. Por ejemplo, cuando tomas foto a una pantalla, a una tela con rayas finas o a una imagen impresa, puedes ver como ondas, curvas o figuras raras que parecen moverse o vibrar. Esto pasa porque los patrones se “interfieren” entre sí o con la resolución del sensor de la cámara, generando ese efecto extraño. No es tanto un error de calidad como el aliasing, sino más bien un efecto de interferencia visual. Es muy importante en fotografía, diseño gráfico y procesamiento de imágenes. Para reducirlo se pueden usar filtros (como el filtro de paso bajo en cámaras), cambiar el ángulo, la distancia o la resolución, o aplicar correcciones digitales.
Conclusión: El moiré es un efecto visual causado por la superposición de patrones que genera distorsiones, y aunque a veces puede ser útil de forma artística, normalmente se busca evitarlo para mantener la imagen clara y fiel a la realidad.
